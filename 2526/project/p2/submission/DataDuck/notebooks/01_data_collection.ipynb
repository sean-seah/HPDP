After test 200 data, start download 5000 data

```python
from tqdm import tqdm
```

```python
# Create Variable
APP_ID = "com.maybank2u.life"

TARGET_REVIEWS = 5000

all_reviews = []

continuation_token = None
```

```python

# Download Reviews in a loop
while len(all_reviews) < TARGET_REVIEWS:

    result, continuation_token = reviews(
        APP_ID,
        lang="en",
        country="my",
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )

    if not result:
        print("No more reviews available.")
        break

    all_reviews.extend(result)

    print(f"Downloaded {len(all_reviews)} reviews...")
```

```python
# Convert to DataFrame
df = pd.DataFrame(all_reviews)

print(df.shape)
```

```python
df = df[
    [
        "reviewId",
        "content",
        "score",
        "at",
        "appVersion"
    ]
]
```

```python
df.columns = [
    "review_id",
    "review",
    "rating",
    "review_date",
    "app_version"
]
```

```python
df = df.drop_duplicates(subset=["review"])
```

```python
print(df.shape)
```

```python
df.to_csv(
    "../data/raw_data/mae_reviews.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset saved successfully!")
```
